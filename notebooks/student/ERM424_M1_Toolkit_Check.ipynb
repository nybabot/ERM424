### Change 1: the filter threshold

The next cell keeps the claims whose total paid is above a threshold and counts them. Change the number 10000 to a different value, then run the cell again. Try 1000, then 50000, and notice how the count moves.

```python
# Filter: keep the claims paid more than THRESHOLD dollars. Change THRESHOLD and run the cell again.
THRESHOLD = 10000
above = losses[losses["total_amount"] > THRESHOLD]
print(len(above), "of", len(losses), "claims paid more than", f"{THRESHOLD:,}", "dollars")
above.sort_values("total_amount", ascending=False).head()
```

### Change 2: the plotted column

The next cell draws a histogram of `total_amount`. Change the column name to one of the other amounts, `injury_amount`, `property_amount` or `vehicle_amount`, then run the cell again and look at how the shape changes.

```python
# Chart: a histogram of one amount column. Change COLUMN and run the cell again.
COLUMN = "total_amount"
fig, ax = plt.subplots(figsize=(7, 3.5))
ax.hist(losses[COLUMN], bins=30)
ax.set_title(f"{COLUMN}: how many claims paid each amount")
ax.set_xlabel(f"{COLUMN} (dollars)")
ax.set_ylabel("number of claims")
plt.show()
```

Figure: histogram of the plotted column. As the notebook arrives it shows `total_amount`: most claims sit in the lowest bars at the left and a few very large claims stretch the axis far to the right. After you change COLUMN, the title and the axis label change with it.
